```python
import tifffile
import numpy as np
import matplotlib.pyplot as plt
import cv2

path = "C:/Users/msgal/Downloads/Ding_Lab/FISH-ML/Composite Images (Multi-Channel)/488_pb5sp4_s1.composite.tif"
img = tifffile.imread(path)
def normalize_channel(channel):
    return cv2.normalize(channel, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

channels = [normalize_channel(img[i]) for i in range(img.shape[0])]

# Display channels
fig, axes = plt.subplots(1, len(channels), figsize=(15, 5))
for i, ch in enumerate(channels):
    axes[i].imshow(ch, cmap='gray')
    axes[i].set_title(f"Channel {i}")
    axes[i].axis("off")
plt.tight_layout()
plt.show()

composite_rgb = np.stack(channels, axis=-1)

# Display composite
plt.figure(figsize=(6, 6))
plt.imshow(composite_rgb)
plt.title("Composite RGB")
plt.axis("off")
plt.show()
```

```python
import tifffile
import numpy as np
import matplotlib.pyplot as plt
import cv2

path = "C:/Users/msgal/Downloads/Ding_Lab/FISH-ML/Composite Images (Multi-Channel)/488_pb5sp4_s1.composite.tif"
img = tifffile.imread(path)

def normalize_channel(channel):
    return cv2.normalize(channel, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

# Normalize channels 0 and 1
ch0 = normalize_channel(img[0])  # e.g., nucleus
ch1 = normalize_channel(img[1])  # e.g., cytoplasm

# Create an RGB composite using ch0 and ch1 only
# For example: ch0→Red, ch1→Green, Blue→Zero
composite_rgb = np.stack([ch0, ch1, np.zeros_like(ch0)], axis=-1)

# Display the composite
plt.figure(figsize=(6, 6))
plt.imshow(composite_rgb)
plt.title("Composite RGB (Ch 0 → R, Ch 1 → G)")
plt.axis("off")
plt.show()
```

```python
import tifffile
import matplotlib.pyplot as plt
import math

img_path = "C:/Users/msgal/Downloads/Ding_Lab/FISH-ML/data1_vadym/multi_channel_tif/EXP_w488_s001.tif"
img = tifffile.imread(img_path)

print(f"Image shape: {img.shape}")

if img.ndim >= 2:
    num_channels = img.shape[0]
    cols = 5  # number of images per row
    rows = math.ceil(num_channels / cols)

    plt.figure(figsize=(15, 3 * rows))
    for i in range(num_channels):
        plt.subplot(rows, cols, i + 1)
        plt.imshow(img[i], cmap="gray")
        plt.title(f"Channel {i}")
        plt.axis("off")

    plt.tight_layout()
    plt.show()
else:
    # Single channel
    plt.imshow(img, cmap="gray")
    plt.title("Single Channel")
    plt.axis("off")
    plt.show()
```
