```python
import numpy as np
import cv2
import tifffile
import matplotlib.pyplot as plt
import pathlib

def normalize_to_uint8(img):
    return cv2.normalize(img, None, 0, 255, cv2.NORM_MINMAX).astype(np.uint8)

def remove_outliers(img, k=15.0, use_median=True):
    """
    Clip values that are more than k std-dev (or MAD units) above center.
    Args:
        img: 16-bit numpy array
        k: threshold (e.g. 3σ)
        use_median: if True use median+MAD, else mean+std
    Returns:
        clipped float32 image in [0,1]
    """
    x = img.astype(np.float32)

    if use_median:
        med = np.median(x)
        mad = np.median(np.abs(x - med)) + 1e-6
        sigma = 1.4826 * mad  # robust std estimate
        thresh = med + k * sigma
    else:
        mean = np.mean(x)
        std = np.std(x)
        thresh = mean + k * std

    # clip outliers
    x_clipped = np.minimum(x, thresh)

    # normalize after clipping (to 0..1)
    x_norm = (x_clipped - x_clipped.min()) / (x_clipped.max() - x_clipped.min() + 1e-6)
    return x_norm


# ----- AN DATA 2 -----

# 1
cyt_path = "C:/Users/msgal/Downloads/Ding_Lab/FISH-ML/data_vadym_singletif_STACKED/MAX_EXP_w488_s0026.tif"
nuc_path = "C:/Users/msgal/Downloads/Ding_Lab/FISH-ML/data_vadym_singletif_STACKED/MAX_EXP_wDAPI_s0026.tif"
cyt_img = tifffile.imread(cyt_path)
nuc_img = tifffile.imread(nuc_path)

smallest = np.amin(cyt_img)
biggest = np.amax(cyt_img)

# Before Removing Outliers
print("An Data 2 Original | Smallest 1:", smallest, "Biggest 1:", biggest)


# ----- OUTLIERS -----

st_dev = 15.0
cyt_removed = remove_outliers(cyt_img, st_dev)

smallest = np.amin(cyt_removed)
biggest = np.amax(cyt_removed)
print("An Data 2 Outliers Removed | Smallest 1:", smallest, "Biggest 1:", biggest)

# ----- NORMALIZED -----

cyt_norm = normalize_to_uint8(cyt_removed)

smallest = np.amin(cyt_norm)
biggest = np.amax(cyt_norm)
print("An Data 2 Outliers Removed + Normalized | Smallest 1:", smallest, "Biggest 1:", biggest)

# ----- VISUALIZATION -----
fig, ax = plt.subplots(nrows=1, ncols=3, figsize=(15, 10), sharex=True, sharey=True)

ax[0].imshow(cyt_img, cmap='gray')
ax[0].set_title("Original")
ax[0].axis('off')

ax[1].imshow(cyt_removed, cmap='gray')
ax[1].set_title(f"Outliers Removed (k={st_dev})")
ax[1].axis('off')

ax[2].imshow(cyt_norm, cmap='gray')
ax[2].set_title(f"Normalized + Outliers Removed (k={st_dev})")
ax[2].axis('off')
```
